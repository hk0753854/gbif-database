```python
dbutils.widgets.text(
    "scientific_name",
    "Hynobius nebulosus",
    "Scientific Name",
)

dbutils.widgets.text(
    "limit",
    "10",
    "Record Limit",
)

scientific_name = dbutils.widgets.get("scientific_name")
limit = int(dbutils.widgets.get("limit"))

print(f"Target species: {scientific_name}")
print(f"Record limit: {limit}")
```

```python
import requests

url = "https://api.gbif.org/v1/occurrence/search"

params = {
    "scientificName": scientific_name,
    "limit": limit,
}

response = requests.get(
    url,
    params=params,
    timeout=30,
)

response.raise_for_status()

data = response.json()

print(f"Scientific name: {scientific_name}")
print(f"Records retrieved: {len(data['results'])}")
```

```python
import json
from datetime import datetime, timezone

from pyspark.sql import Row


ingested_at = datetime.now(timezone.utc)

rows = [
    Row(
        raw_json=json.dumps(record, ensure_ascii=False),
        query_scientific_name=scientific_name,
        ingested_at=ingested_at,
    )
    for record in data["results"]
]

df_bronze = spark.createDataFrame(rows)
```

```python
%skip
%sql
TRUNCATE TABLE workspace.bronze.gbif_occurrences
```

```python
bronze_table = "workspace.bronze.gbif_occurrences"

(
    df_bronze.write
    .format("delta")
    .mode("append")
    .saveAsTable(bronze_table)
)

print(
    f"Bronze table updated: {bronze_table}, "
    f"records={len(rows)}, "
    f"scientific_name={scientific_name}"
)
```
